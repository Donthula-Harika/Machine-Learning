## Hypothesis Testing: Tip vs Taxi Color

### Need for Hypothesis Testing
- Visual analysis shows differences in tips between yellow and green taxis.
- However, graphical analysis alone is not sufficient to make a conclusion.
- Hypothesis testing is used to statistically verify whether taxi color affects tip amount.


### Test Used
- Independent T-Test
- Used to compare the mean tip values of two independent groups:
  - Yellow taxis
  - Green taxis


### Hypotheses
- **Null Hypothesis (H₀):** Mean tip of yellow taxis = Mean tip of green taxis
- **Alternative Hypothesis (H₁):** Mean tip of yellow taxis ≠ Mean tip of green taxis


```python
yellow_tip  = df['tip'].loc[df['color']=='yellow']
green_tip = df['tip'].loc[df['color']=='green']
```

```python
yellow_tip.shape
```

```python
green_tip.shape
```

```python
from scipy import stats

t_stat, p_val = stats.ttest_ind(yellow_tip, green_tip)

print("Independent T-Test statistic:", t_stat)
print("P-value:", p_val)
if p_val < 0.05:
    print("Reject null hypothesis: Tips differ by taxi color")
else:
    print("Fail to reject null hypothesis: No significant tip difference by taxi color")
```

### Conclusion
- The p-value is less than 0.05.
- Null hypothesis is rejected.
- Taxi color has a statistically significant effect on tip amount.

## Hypothesis Testing: Tip vs Payment Method

### Objective
- To determine whether payment method has a significant effect on tip amount.

### Hypotheses
- **H₀:** Distribution of tips for cash payments is equal to distribution of tips for credit card payments.
- **H₁:** Distribution of tips for cash payments is not equal to distribution of tips for credit card payments.

### Test Used
- Independent T-Test (two independent groups, numerical target variable).

### Procedure
- Tips are divided into two independent samples based on payment method.
- Independent T-Test is applied to compare mean tip values.




```python
card = df['tip'].loc[df['payment'] == 'credit card']
cash = df['tip'].loc[df['payment'] == 'cash']
```

```python
card.shape
```

```python
cash.shape
```

```python
from scipy import stats
t_stat, p_val = stats.ttest_ind(card, cash)

print("Independent T-Test statistic:", t_stat)
print("P-value:", p_val)

if p_val < 0.05:
    print("Reject null hypothesis: Tips differ by payment type")
else:
    print("Fail to reject null hypothesis: No significant tip difference by payment type")
```

### Observation
- The p-value obtained from the Independent T-Test is 0.0, which is less than the significance level of 0.05.
- This indicates a statistically significant difference in tip amounts between cash and credit card payments.

### Conclusion
- The null hypothesis is rejected.
- Payment method has a significant effect on tip amount.
- Tips differ significantly between cash and credit card payment types.

## Hypothesis Testing: Tip vs Pickup Borough (ANOVA)

### Objective
- To determine whether the **mean tip amount differs across pickup boroughs**.

### Why ANOVA is Used
- Tip is a numerical variable.
- Pickup borough is a categorical variable with **more than two groups**:
  - Manhattan
  - Brooklyn
  - Queens
  - Bronx
- ANOVA (Analysis of Variance) is used to compare **means across multiple independent groups**.


### Hypotheses
- **H₀:** Mean tip is the same across all pickup boroughs.
- **H₁:** At least one pickup borough has a different mean tip.


### Data Preparation
- Tips are separated based on pickup borough to form independent samples.


```python
mannhattan = df['tip'].loc[df['pickup_borough']=='Manhattan']
brooklyn = df['tip'].loc[df['pickup_borough']=='Brooklyn']  
queens = df['tip'].loc[df['pickup_borough']=='Queens']
bronx = df['tip'].loc[df['pickup_borough']=='Bronx']
```

### Performing One-Way ANOVA


```python
from scipy import stats

f_stat, p_val = stats.f_oneway(mannhattan, brooklyn, queens, bronx)
print("ANOVA F-statistic:", f_stat)
print("P-value:", p_val)
```

### Decision Rule
- Significance level (α) = 0.05  
- If *p-value < 0.05* → Reject H₀  
- If *p-value ≥ 0.05* → Fail to reject H₀

```python
if p_val < 0.05:
    print("Reject null hypothesis:")
else:
    print("Fail to reject null hypothesis: No significant difference by Region")
```

### Interpretation
- ANOVA helps determine whether **pickup borough influences tip amount statistically**, beyond visual analysis.


### Observation
- The p-value obtained from the ANOVA test is extremely small (≈ 5.15 × 10⁻⁵⁸), which is far below the significance level of 0.05.
- This indicates strong statistical evidence that mean tip values differ across pickup boroughs.

### Conclusion
- The null hypothesis is rejected.
- There is a statistically significant difference in mean tip amounts among the pickup boroughs.
- Pickup borough has a significant effect on the tip amount.


## Post-hoc Analysis: Tukey HSD Test (Pickup Borough vs Tip)

### Why Tukey HSD Test is Used
- ANOVA confirmed that mean tip amounts differ across pickup boroughs.
- However, ANOVA does not indicate **which specific borough pairs differ**.
- Tukey HSD is a post-hoc test used to identify **pairwise differences** between groups.


### Hypotheses (for each pair of boroughs)
- **H₀:** Mean tip of Borough A = Mean tip of Borough B
- **H₁:** Mean tip of Borough A ≠ Mean tip of Borough B


### Performing Tukey HSD Test


```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Prepare data
tip = df['tip']
pickup_borough = df['pickup_borough']

# Perform Tukey HSD
tukey_result = pairwise_tukeyhsd(endog=tip, groups=pickup_borough, alpha=0.05)

print(tukey_result)
```

### Observation (Tukey HSD Test Results)

- All borough pair comparisons show **reject = True**, indicating statistically significant differences in mean tip values.
- Bronx has significantly lower mean tips compared to Brooklyn, Manhattan, and Queens.
- Brooklyn has significantly lower mean tips than Manhattan and Queens.
- Manhattan has significantly lower mean tips compared to Queens.
- Queens consistently shows the **highest mean tip values** among all pickup boroughs.


### Conclusion
- The Tukey HSD post-hoc analysis confirms that **mean tip amounts differ significantly between every pair of pickup boroughs**.
- Queens receives the highest tips, while Bronx receives the lowest.
- This validates the ANOVA result and confirms that **pickup borough strongly influences tipping behavior**.
