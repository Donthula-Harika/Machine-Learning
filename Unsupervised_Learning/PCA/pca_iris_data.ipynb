```python
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

```

```python
iris = load_iris()
```

```python
X =iris.data
y = iris.target
```

```python
X_scaled = StandardScaler().fit_transform(X)
```

```python
X.shape
```

### APPLY PCA and Visulization 

```python
# APPLY PCA
pca = PCA()
X_pca  = pca.fit_transform(X_scaled)

#Screen plot
plt.figure(figsize=(8,5))
plt.plot(range(1, len(pca.explained_variance_ratio_)+1), pca.explained_variance_ratio_, 'o-', label = 'Individual Explained Variance')
plt.plot(range(1, len(pca.explained_variance_ratio_)+1), np.cumsum(pca.explained_variance_ratio_), 's--', label = 'Cumulative Explained Variance')
plt.xlabel ("Number of Components")
plt.ylabel('Explained Variance')
plt.title('Screen plot')
plt.grid(True)
plt.legend()
plt.show()

```

### Choose NO.of Components (95% variance)

```python
pca_95 = PCA(n_components=0.95)

X_reduced = pca_95.fit_transform(X_scaled)
print(f"Reduced Shape: {X_reduced.shape}")
```

### Visualize top 2 components

```python
pca_2d  =PCA(n_components=2)
X_2d = pca_2d.fit_transform(X_scaled)

plt.figure(figsize=(8,6))
sns.scatterplot(x=X_2d[:,0],y=X_2d[:,1], hue = iris.target_names[y],palette='Set1')

plt.xlabel ("PC1")
plt.ylabel('PC2')
plt.title('2D Projrction of Iris Dataset')
plt.grid(True)
plt.legend()
plt.show()
```
