```python
import pandas as pd
```

```python
#load data
df = pd.read_csv("Position_Salaries.csv")
df.head()
```

```python
X = df[["Level"]]  
y = df["Salary"]   
```

```python
# Training the Linear Regression model on the whole dataset
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, y)
 
# Training the Polynomial Regression model on the whole dataset
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree = 4)
X_poly = poly_reg.fit_transform(X)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X_poly, y)
 
print("✅ Models Trained Successfully!")
 
 
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler    

# Visualising the Polynomial Regression results (for higher resolution and smoother curve)
import numpy as np  
X_grid = np.arange(min(X.values), max(X.values), 0.1)  # 0.1 step size
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color = 'blue')
plt.plot(X_grid, lin_reg_2.predict(poly_reg.fit_transform(X_grid)), color = 'red')
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
# Predicting a new result with Linear Regression
y_pred = lin_reg.predict([[6.5]])
print(f"Predicted salary using Linear Regression for level 6.5: {y_pred}")
# Predicting a new result with Polynomial Regression
y_pred_poly = lin_reg_2.predict(poly_reg.fit_transform([[6.5]]))
print(f"Predicted salary using Polynomial Regression for level 6.5: {y_pred_poly}") 

                                                                     
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(8,5))

# Actual points
plt.scatter(X, y, color='red')

# Linear Regression line
plt.plot(X, lin_reg.predict(X), color='blue')

plt.title("Truth or Bluff (Linear Regression)")
plt.xlabel("Position Level")
plt.ylabel("Salary")
plt.show()
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Predictions on entire dataset
y_pred = lin_reg.predict(X)

# Metrics
mse = mean_squared_error(y, y_pred)
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

# Print metrics
print("Mean Squared Error (MSE):", mse)
print("Mean Absolute Error (MAE):", mae)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

```python
# Polynomial Regression metrics
y_poly_pred = lin_reg_2.predict(poly_reg.transform(X))

mse_p = mean_squared_error(y, y_poly_pred)
mae_p = mean_absolute_error(y, y_poly_pred)
rmse_p = np.sqrt(mse_p)
r2_p = r2_score(y, y_poly_pred)

print("Polynomial MSE:", mse_p)
print("Polynomial MAE:", mae_p)
print("Polynomial RMSE:", rmse_p)
print("Polynomial R²:", r2_p)
```

```python
# Linear Regression prediction
linear_prediction = lin_reg.predict([[6.5]])
print("Linear Regression Prediction for Level 6.5:", linear_prediction)

# Polynomial Regression prediction
poly_prediction = lin_reg_2.predict(poly_reg.transform([[6.5]]))
print("Polynomial Regression Prediction for Level 6.5:", poly_prediction)
```
