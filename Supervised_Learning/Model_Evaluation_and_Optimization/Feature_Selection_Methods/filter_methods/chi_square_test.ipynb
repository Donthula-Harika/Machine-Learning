```python
# You are building a classification model to predict whether a loan will be approved.
# Dataset features:
# Gender (Male/Female)
# Marital Status (Yes/No)
# Education Level (Graduate/Non-Graduate)
# Applicant Income
# Property Area (Urban/Rural)
# Favorite Color
# Loan Status (Approved/Rejected)
# You want to identify which categorical features significantly influence loan approval.

import pandas as pd
# Feature matrix (X)
x = pd.DataFrame({
    'gender':   [1, 0, 1, 0, 1],   # 1-Male, 0-Female
    'married':  [1, 1, 0, 1, 0],   # 1-Married, 0-Not Married
    'education':[1, 1, 0, 1, 0],   # 1-Graduate, 0-Non-Graduate
    'property': [1, 0, 1, 1, 0],   # 1-Urban, 0-Rural
    'fav_color':[0, 1, 0, 1, 0]    # Random / Irrelevant feature
})

# Target variable (y)
y = pd.Series([1, 0, 1, 1, 0], name="loan_status")  # 1-Approved, 0-Rejected

x, y

```

```python
# Chi-Square Requires Non-Negative values

from sklearn.feature_selection import chi2,SelectKBest
# from sklearn.feature_extraction import SelectKBest
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(x)

selector = SelectKBest(score_func=chi2,k=2)
X_new = selector.fit_transform(X_scaled,y)

selected_features= x.columns[selector.get_support()]

print("Selected Features: ", selected_features ) 

```
